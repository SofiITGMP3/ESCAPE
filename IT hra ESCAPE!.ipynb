```python


def play_game():
    print ("Vítej ve hře ESCAPE!") 
    print ("Stojíš ve vstupní hale strašidelného domu.")
    print("Hlavní dveře jsou zamčené a dveře do dalšího pokoje mají kódový zámek.")
    print("Vyber si, jestli chceš prohledat pokoj (stiskni a) nebo zadat kód (stiskni b).")
    pokusy = 0
    zivoty = 3


    while pokusy <3:
        volba = input("Zvol si a nebo b")
        print (volba)

        if volba == "a":
            print("Díváš se okolo a na stěně visí staré kyvadlové hodiny. Ukazují čas 12:00")
            kod = input("Zadej kód, který je čtyřmístný a číselný.")
            if kod == "1200":
                print("Dveře se otevřely a pokračuješ do laboratoře.")
                break
            if kod != "1200":
                print("Zadal jsi špatný kód a přišel si o 1 pokus, máš celkem 3 pokusy. Stiskni a pro prohledání pokoje nebo stiskni b pro zadání kódu.")
                pokusy += 1
                print(pokusy)

            

        if volba =="b":
            kod = input("Zadej kód, který je čtyřmístný a číselný.")
            if kod == "1200":
                print("Dveře se otevřely a pokračuješ do laboratoře.")            
                break
            if kod !="1200":
                print("Zadal jsi špatný kód a přišel si o 1 pokus, máš celkem 3 pokusy. Stiskni a pro prohledání pokoje nebo stiskni b pro zadání kódu.")
                pokusy += 1
                print(pokusy)

        if volba != "a" and volba != "b":
            print ("Zadal jsi špatné písmeno. Přišel si o 1 pokus, máš celkem 3 pokusy. Stiskni a pro prohledání pokoje nebo stiskni b pro zadání kódu.")
            pokusy += 1
    if pokusy == 3:
        print("Hra končí")
        odpoved = input("Chceš hrát hru? (ano/ne): ")
        if odpoved.lower() == "ano":
            play_game()
        else:
            print("good bye")
        return

    def pokoje():
        nonlocal zivoty
        print("vyber si postupně 3 předměty")
        print("Leží tu lékárnička, česnek, baterka, šátek, jídlo")
        batoh=[]

        for i in range (0,3):
            predmet = input ("napiš si 1 předmět, který si chceš vzít (celým slovem)")
            if predmet.lower() == "lékárnička" or predmet.lower() == "lekarnicka" :
                batoh.append("lékárnička")
            elif predmet.lower() == "česnek" or predmet.lower() == "cesnek": 
                batoh.append("česnek") 
            elif predmet.lower() == "baterka":
                batoh.append("baterka") 
            elif predmet.lower() == "šátek" or predmet.lower() == "satek":
                batoh.append("šátek") 
            elif  predmet.lower() == "jídlo" or predmet.lower() == "jidlo":
                batoh.append("jídlo")   
            else:    
                print("Napsal jsi to špatně, teď jsi můžeš vzít o 1 věc méně")

        print("V batůžku máš:")
        print(batoh)
        print("a pokračuješ do sklepa.")
        print("Vzbudil jsi upíra, vylezl z dřevěné rakve a jde za tebou.")
        print("Vyber si z batohu, co ti může pomoci proti upírovi.")
        vec = input("zapiš, co chceš použít proti upírovi:")
        
        if vec.lower() == "česnek" or vec.lower() == "cesnek":
            if "česnek" in batoh:
                print("upír utekl")
                print("Podíval jsi se do rakve a vidíš tam klíč. Běžíš s ním ke vstupním dveřím a odemykáš. VYHRÁL JSI!!!")
            else:
                print("Nemáš česnek v batohu.") 
                print("Upír se ti zakousl do krku, sebral ti věci z batohu a ztrácíš 1 život.")
                zivoty -= 1
                print("počet životů: %i" %zivoty)
                if zivoty > 0:
                    print("Vracíš se do laboratoře a chceš si znovu naplnit batoh.")
                    pokoje()
                else:
                    print("Došly ti životy. Hra končí!")
            
                   
            
        else:
            print("Tohoto předmětu se upír nebojí. Zakousl se ti do krku, sebral ti věci z batohu a ztrácíš 1 život.")
            zivoty -= 1
            print("počet životů: %i" %zivoty)
            if zivoty > 0:
                print("Vracíš se do laboratoře a chceš si znovu naplnit batoh.")
                pokoje()
            else:
                print("Došly ti životy. Hra končí!")    
            
    pokoje()    
#play_game()
odpoved = input("Chceš hrát hru? (ano/ne): ")
if odpoved.lower() == "ano":
    play_game()
else:
    print("good bye")        
```
